```python
key = {'a':1,
        'b':2,
        'c':3,
        'd':4,
        'e':5,
        'f':6,
        'g':7,
        'h':8,
        'i':9,
        'j':10,
        'k':11,
        'l':12,
        'm':13,
        'n':14,
        'o':15,
        'p':16,
        'q':17,
        'r':18,
        's':19,
        't':20,
        'u':21,
        'v':22,
        'w':23,
        'x':24,
        'y':25,
        'z':26}
```

```python
import re
```

```python
def obtain_numerical_core(input: str, show_details: bool = False):
    """
    Takes a 4-letter word, encodes it as numbers, then calculates the numerical core (according to the lore established in Blue Prince)

    Numerical cores
        x
    
    Args:
        - input: 4-letter word to be encoded
        - show_details: bool to determine if calculation details should be printed or not
    
    Returns:
        - The numerical core of the encoded input word
    """
    # Ensure the input is a 4-letter word
    if isinstance(input, str):
        if len(input) == 4:
            if bool(re.search(r'\d', input)) == False:
                # Break into characters
                    characters = list(input)
                
                    # Map to the numerical values
                    numbers = []
                    for char in characters:
                        numbers = numbers + [key[char]]
                
                    # Calculate all possible results
                    results = []
                    results = results + [(((numbers[0] - numbers[1]) / numbers[2]) * numbers[3])]
                    results = results + [(((numbers[0] - numbers[1]) * numbers[2]) / numbers[3])]
                    results = results + [(((numbers[0] * numbers[1]) - numbers[2]) / numbers[3])]
                    results = results + [(((numbers[0] * numbers[1]) / numbers[2]) - numbers[3])]
                    results = results + [(((numbers[0] / numbers[1]) * numbers[2]) - numbers[3])]
                    results = results + [(((numbers[0] / numbers[1]) - numbers[2]) * numbers[3])]
                
                    # Filter to positive integers
                    final_results = []
                    for result in results:
                        # if result is a positive integer
                        if result > 0 and result == int(result):
                            final_results = final_results + [result]
                    if len(final_results) == 0:
                        print(f"‼️Error: It is not possible to calculate a numberical core for '{input}' (there are no positive integer results)")
                
                    if show_details is True:
                        print("Individual characters:")
                        print(f"\t{characters}\n")
                        print("Mapped to numbers:")
                        print(f"\t{numbers}\n")
                        print("All possible results:")
                        print(f"\t{results}\n")
                        print("Valid (positive integer) results:")
                        print(f"\t{final_results}\n")
                
                    print("Final result:")
                    print(f"\t{min(final_results)}")
                
                    # Return the smallest valid result
                    return min(final_results)                
            else:
                print(f"‼️Error: input ({input}) contains a number; it must only contain letters")
        else:
            print(f"‼️Error: input ({input}) is not exactly 4 letters long")
    else:
        print(f"‼️Error: input ({input}) is not a string")
```
